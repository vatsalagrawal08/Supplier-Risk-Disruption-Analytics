```python
import pandas as pd

# Load the datasets
parts = pd.read_csv("data/raw/parts_master.csv")
purchase_orders = pd.read_csv("data/processed/purchase_orders_processed.csv")
supplier_risk = pd.read_csv("data/processed/supplier_risk_assessment.csv")

# Inspect dimensions
print("PARTS MASTER")
print(parts.shape)
print(parts.columns.tolist())
print()

print("PURCHASE ORDERS")
print(purchase_orders.shape)
print(purchase_orders.columns.tolist())
print()

print("SUPPLIER RISK ASSESSMENT")
print(supplier_risk.shape)
print(supplier_risk.columns.tolist())
```

```python
# Inspect supplier-part relationships

print("Unique suppliers in parts master:")
print(parts["supplier_id_primary"].nunique())

print("\nUnique suppliers in purchase orders:")
print(purchase_orders["supplier_id"].nunique())

print("\nUnique parts in parts master:")
print(parts["part_id"].nunique())

print("\nUnique parts in purchase orders:")
print(purchase_orders["part_id"].nunique())
```

```python
# Sample records to understand the actual structure

display(parts.head())
display(purchase_orders.head())
display(supplier_risk.head())
```

```python
# View suppliers ranked by final priority score

supplier_risk_ranked = supplier_risk.sort_values(
    "supplier_priority_score",
    ascending=False
)

display(
    supplier_risk_ranked[
        [
            "supplier_id",
            "supplier_priority_score",
            "risk_category",
            "performance_risk_score",
            "business_criticality_score"
        ]
    ]
)
```

```python
# Select suppliers that deserve disruption analysis

disruption_candidates = supplier_risk[
    supplier_risk["risk_category"].isin(["Critical", "High"])
].sort_values(
    "supplier_priority_score",
    ascending=False
)

display(
    disruption_candidates[
        [
            "supplier_id",
            "supplier_priority_score",
            "risk_category",
            "performance_risk_score",
            "business_criticality_score"
        ]
    ]
)
```

```python
print("Disruption candidates:")
print("Critical:", (disruption_candidates["risk_category"] == "Critical").sum())
print("High:", (disruption_candidates["risk_category"] == "High").sum())
print("Total:", len(disruption_candidates))
```

```python
# Map selected disruption candidates to the parts they have actually supplied

supplier_part_history = (
    purchase_orders[
        purchase_orders["supplier_id"].isin(
            disruption_candidates["supplier_id"]
        )
    ]
    .groupby(["supplier_id", "part_id"])
    .agg(
        total_orders=("po_id", "count"),
        total_order_quantity=("ordered_qty", "sum"),
        total_received_quantity=("received_qty", "sum")
    )
    .reset_index()
)

# Add part information
supplier_part_history = supplier_part_history.merge(
    parts[
        [
            "part_id",
            "part_family",
            "criticality_class",
            "unit_cost",
            "supplier_id_primary"
        ]
    ],
    on="part_id",
    how="left"
)

# Sort by supplier and criticality
supplier_part_history = supplier_part_history.sort_values(
    ["supplier_id", "criticality_class", "total_order_quantity"],
    ascending=[True, True, False]
)

display(supplier_part_history.head(20))
```

```python
# Summarize the parts exposed under each supplier disruption

supplier_disruption_summary = (
    supplier_part_history
    .groupby("supplier_id")
    .agg(
        affected_parts=("part_id", "nunique"),
        critical_parts_affected=(
            "criticality_class",
            lambda x: (x == "A").sum()
        ),
        total_order_quantity_exposed=(
            "total_order_quantity",
            "sum"
        )
    )
    .reset_index()
)

# Add supplier risk information
supplier_disruption_summary = supplier_disruption_summary.merge(
    supplier_risk[
        [
            "supplier_id",
            "supplier_priority_score",
            "risk_category"
        ]
    ],
    on="supplier_id",
    how="left"
)

supplier_disruption_summary = supplier_disruption_summary.sort_values(
    "supplier_priority_score",
    ascending=False
)

display(supplier_disruption_summary)
```

```python
# Find all suppliers that have historically supplied each affected part

part_supplier_history = (
    purchase_orders
    .groupby("part_id")
    .agg(
        supplier_count=("supplier_id", "nunique"),
        suppliers=("supplier_id", lambda x: sorted(x.unique()))
    )
    .reset_index()
)

# Attach this information to our disruption-level supplier-part data
supplier_part_alternatives = supplier_part_history.merge(
    part_supplier_history,
    on="part_id",
    how="left"
)

# A part is single-source if only one supplier has historically supplied it
supplier_part_alternatives["is_single_source"] = (
    supplier_part_alternatives["supplier_count"] == 1
)

display(
    supplier_part_alternatives[
        [
            "supplier_id",
            "part_id",
            "part_family",
            "criticality_class",
            "total_order_quantity",
            "supplier_count",
            "suppliers",
            "is_single_source"
        ]
    ].head(30)
)
```

```python
# Summarize alternative-supplier availability for each disruption scenario

alternative_supplier_summary = (
    supplier_part_alternatives
    .groupby("supplier_id")
    .agg(
        affected_parts=("part_id", "nunique"),
        parts_with_alternatives=(
            "is_single_source",
            lambda x: (~x).sum()
        ),
        single_source_parts=(
            "is_single_source",
            "sum"
        ),
        critical_single_source_parts=(
            "is_single_source",
            lambda x: (
                x[
                    supplier_part_alternatives.loc[x.index, "criticality_class"] == "A"
                ]
            ).sum()
        )
    )
    .reset_index()
)

# Add risk information
alternative_supplier_summary = alternative_supplier_summary.merge(
    supplier_risk[
        [
            "supplier_id",
            "supplier_priority_score",
            "risk_category"
        ]
    ],
    on="supplier_id",
    how="left"
)

alternative_supplier_summary = alternative_supplier_summary.sort_values(
    "supplier_priority_score",
    ascending=False
)

display(alternative_supplier_summary)
```

```python
supply_chain_history = pd.read_csv(
    "data/raw/supply_chain_history.csv"
)

print("Supply Chain History shape:", supply_chain_history.shape)
print("\nColumns:")
print(supply_chain_history.columns.tolist())

display(supply_chain_history.head())
```

```python
print(
    "Date range:",
    supply_chain_history["date"].min(),
    "to",
    supply_chain_history["date"].max()
)
```

```python
# Aggregate operational exposure for parts affected by each disruption candidate

part_operational_exposure = (
    supply_chain_history
    .groupby("part_id")
    .agg(
        total_consumption=("consumption_qty", "sum"),
        avg_daily_consumption=("consumption_qty", "mean"),
        avg_on_hand=("on_hand_qty", "mean"),
        min_on_hand=("on_hand_qty", "min"),
        total_backorder=("backorder_qty", "sum"),
        max_backorder=("backorder_qty", "max"),
        total_blocked=("blocked_qty", "sum")
    )
    .reset_index()
)

# Attach operational information to affected supplier-part combinations
supplier_part_exposure = supplier_part_history.merge(
    part_operational_exposure,
    on="part_id",
    how="left"
)

# Add unit cost
supplier_part_exposure["procurement_value_exposed"] = (
    supplier_part_exposure["total_order_quantity"]
    * supplier_part_exposure["unit_cost"]
)

display(
    supplier_part_exposure[
        [
            "supplier_id",
            "part_id",
            "part_family",
            "criticality_class",
            "total_order_quantity",
            "unit_cost",
            "procurement_value_exposed",
            "total_consumption",
            "avg_daily_consumption",
            "avg_on_hand",
            "total_backorder",
            "total_blocked"
        ]
    ].head(20)
)
```

```python
# Add single-source information to the exposure dataset

supplier_part_exposure = supplier_part_exposure.merge(
    supplier_part_alternatives[
        [
            "supplier_id",
            "part_id",
            "is_single_source"
        ]
    ],
    on=["supplier_id", "part_id"],
    how="left"
)

print("Columns added successfully:")
print(
    supplier_part_exposure[
        ["supplier_id", "part_id", "is_single_source"]
    ].head()
)
```

```python
# Aggregate part-level exposure into supplier disruption scenarios

disruption_scenario_summary = (
    supplier_part_exposure
    .groupby("supplier_id")
    .agg(
        affected_parts=("part_id", "nunique"),

        critical_parts_affected=(
            "criticality_class",
            lambda x: (x == "A").sum()
        ),

        total_order_quantity_exposed=(
            "total_order_quantity",
            "sum"
        ),

        critical_order_quantity_exposed=(
            "total_order_quantity",
            lambda x: x[
                supplier_part_exposure.loc[
                    x.index, "criticality_class"
                ] == "A"
            ].sum()
        ),

        procurement_value_exposed=(
            "procurement_value_exposed",
            "sum"
        ),

        critical_procurement_value_exposed=(
            "procurement_value_exposed",
            lambda x: x[
                supplier_part_exposure.loc[
                    x.index, "criticality_class"
                ] == "A"
            ].sum()
        ),

        total_consumption=(
            "total_consumption",
            "sum"
        ),

        average_daily_consumption=(
            "avg_daily_consumption",
            "sum"
        ),

        average_on_hand_inventory=(
            "avg_on_hand",
            "sum"
        ),

        total_backorder=(
            "total_backorder",
            "sum"
        ),

        total_blocked_quantity=(
            "total_blocked",
            "sum"
        ),

        single_source_parts=(
            "is_single_source",
            "sum"
        )
    )
    .reset_index()
)

# Add supplier risk information
disruption_scenario_summary = disruption_scenario_summary.merge(
    supplier_risk[
        [
            "supplier_id",
            "supplier_priority_score",
            "risk_category",
            "performance_risk_score",
            "business_criticality_score"
        ]
    ],
    on="supplier_id",
    how="left"
)

# Sort by supplier priority
disruption_scenario_summary = disruption_scenario_summary.sort_values(
    "supplier_priority_score",
    ascending=False
)

display(disruption_scenario_summary)
```

```python
# Business-facing disruption scenario table

disruption_business_view = disruption_scenario_summary[
    [
        "supplier_id",
        "risk_category",
        "supplier_priority_score",
        "affected_parts",
        "critical_parts_affected",
        "total_order_quantity_exposed",
        "critical_order_quantity_exposed",
        "procurement_value_exposed",
        "critical_procurement_value_exposed",
        "total_consumption",
        "average_on_hand_inventory",
        "total_backorder",
        "total_blocked_quantity",
        "single_source_parts"
    ]
].copy()

display(disruption_business_view)
```

```python
# Calculate network-level daily consumption for each disruption scenario

total_days = supply_chain_history["date"].nunique()

scenario_consumption = (
    supplier_part_exposure[
        ["supplier_id", "part_id"]
    ]
    .drop_duplicates()
    .merge(
        supply_chain_history[
            ["part_id", "date", "site_id", "consumption_qty"]
        ],
        on="part_id",
        how="left"
    )
    .groupby("supplier_id")
    .agg(
        total_consumption=("consumption_qty", "sum")
    )
    .reset_index()
)

scenario_consumption["average_daily_consumption"] = (
    scenario_consumption["total_consumption"] / total_days
)

print("Number of days in supply-chain history:", total_days)

display(scenario_consumption)
```

```python
# Replace the earlier daily-consumption calculation
disruption_scenario_summary = disruption_scenario_summary.drop(
    columns=["total_consumption", "average_daily_consumption"],
    errors="ignore"
)

disruption_scenario_summary = disruption_scenario_summary.merge(
    scenario_consumption,
    on="supplier_id",
    how="left"
)

disruption_scenario_summary = disruption_scenario_summary.sort_values(
    "supplier_priority_score",
    ascending=False
)

display(
    disruption_scenario_summary[
        [
            "supplier_id",
            "risk_category",
            "supplier_priority_score",
            "affected_parts",
            "critical_parts_affected",
            "total_order_quantity_exposed",
            "critical_order_quantity_exposed",
            "procurement_value_exposed",
            "critical_procurement_value_exposed",
            "total_consumption",
            "average_daily_consumption",
            "total_backorder",
            "total_blocked_quantity",
            "single_source_parts"
        ]
    ]
)
```

```python
# ============================================================
# DISRUPTION EXPOSURE RANKING
# ============================================================

exposure_ranking = disruption_scenario_summary[
    [
        "supplier_id",
        "risk_category",
        "supplier_priority_score",
        "affected_parts",
        "critical_parts_affected",
        "procurement_value_exposed",
        "critical_procurement_value_exposed",
        "average_daily_consumption",
        "total_backorder",
        "total_blocked_quantity"
    ]
].copy()

# Rank each disruption consequence.
# Higher exposure = better rank (Rank 1 = highest exposure)

exposure_ranking["rank_critical_parts"] = (
    exposure_ranking["critical_parts_affected"]
    .rank(method="min", ascending=False)
)

exposure_ranking["rank_critical_procurement"] = (
    exposure_ranking["critical_procurement_value_exposed"]
    .rank(method="min", ascending=False)
)

exposure_ranking["rank_procurement"] = (
    exposure_ranking["procurement_value_exposed"]
    .rank(method="min", ascending=False)
)

exposure_ranking["rank_daily_consumption"] = (
    exposure_ranking["average_daily_consumption"]
    .rank(method="min", ascending=False)
)

exposure_ranking["rank_backorder"] = (
    exposure_ranking["total_backorder"]
    .rank(method="min", ascending=False)
)

exposure_ranking["rank_blocked"] = (
    exposure_ranking["total_blocked_quantity"]
    .rank(method="min", ascending=False)
)

# Equal-weight rank aggregation
rank_columns = [
    "rank_critical_parts",
    "rank_critical_procurement",
    "rank_procurement",
    "rank_daily_consumption",
    "rank_backorder",
    "rank_blocked"
]

exposure_ranking["average_exposure_rank"] = (
    exposure_ranking[rank_columns].mean(axis=1)
)

# Rank the final disruption exposure
exposure_ranking["disruption_exposure_rank"] = (
    exposure_ranking["average_exposure_rank"]
    .rank(method="min", ascending=True)
    .astype(int)
)

# Final business view
disruption_exposure_ranking = exposure_ranking.sort_values(
    "disruption_exposure_rank"
)

display(
    disruption_exposure_ranking[
        [
            "disruption_exposure_rank",
            "supplier_id",
            "risk_category",
            "supplier_priority_score",
            "affected_parts",
            "critical_parts_affected",
            "procurement_value_exposed",
            "critical_procurement_value_exposed",
            "average_daily_consumption",
            "total_backorder",
            "total_blocked_quantity",
            "average_exposure_rank"
        ]
    ]
)
```

```python
# ============================================================
# SAVE FINAL DISRUPTION SCENARIO ANALYSIS
# ============================================================

# Start from the complete scenario summary
final_disruption_output = disruption_scenario_summary.merge(
    exposure_ranking[
        [
            "supplier_id",
            "disruption_exposure_rank",
            "average_exposure_rank"
        ]
    ],
    on="supplier_id",
    how="left"
)

# Select the final business-facing columns
output_columns = [
    "disruption_exposure_rank",
    "supplier_id",
    "risk_category",
    "supplier_priority_score",
    "performance_risk_score",
    "business_criticality_score",
    "affected_parts",
    "critical_parts_affected",
    "total_order_quantity_exposed",
    "critical_order_quantity_exposed",
    "procurement_value_exposed",
    "critical_procurement_value_exposed",
    "total_consumption",
    "average_daily_consumption",
    "average_on_hand_inventory",
    "total_backorder",
    "total_blocked_quantity",
    "single_source_parts",
    "average_exposure_rank"
]

final_disruption_output = (
    final_disruption_output[output_columns]
    .sort_values("disruption_exposure_rank")
)

# Save final disruption scenario analysis
output_path = "data/processed/disruption_scenario_analysis.csv"

final_disruption_output.to_csv(
    output_path,
    index=False
)

print("Saved successfully:")
print(output_path)

display(final_disruption_output.head(10))
```
