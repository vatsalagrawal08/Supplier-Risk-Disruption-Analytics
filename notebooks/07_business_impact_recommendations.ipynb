```python
import pandas as pd

# Load finalized disruption analysis
disruption = pd.read_csv(
    "data/processed/disruption_scenario_analysis.csv"
)

print("Shape:", disruption.shape)
print("\nColumns:")
print(disruption.columns.tolist())

display(disruption.head(10))
```

```python
# Focus on the most important disruption drivers

recommendation_view = disruption[
    [
        "disruption_exposure_rank",
        "supplier_id",
        "risk_category",
        "supplier_priority_score",
        "affected_parts",
        "critical_parts_affected",
        "total_order_quantity_exposed",
        "critical_order_quantity_exposed",
        "procurement_value_exposed",
        "critical_procurement_value_exposed",
        "average_daily_consumption",
        "total_backorder",
        "total_blocked_quantity",
        "single_source_parts"
    ]
].sort_values(
    "disruption_exposure_rank"
)

display(recommendation_view)
```

```python
# ============================================================
# RECOMMENDATION DRIVERS
# ============================================================

recommendation_drivers = disruption.copy()

# Data-driven thresholds
critical_parts_threshold = recommendation_drivers[
    "critical_parts_affected"
].quantile(0.75)

procurement_threshold = recommendation_drivers[
    "procurement_value_exposed"
].quantile(0.75)

consumption_threshold = recommendation_drivers[
    "average_daily_consumption"
].quantile(0.75)

blocked_threshold = recommendation_drivers[
    "total_blocked_quantity"
].quantile(0.75)

backorder_threshold = recommendation_drivers[
    "total_backorder"
].quantile(0.75)

print("Data-driven thresholds:")
print("Critical parts:", critical_parts_threshold)
print("Procurement exposure:", procurement_threshold)
print("Daily consumption:", consumption_threshold)
print("Blocked quantity:", blocked_threshold)
print("Backorder:", backorder_threshold)
```

```python
# ============================================================
# FLAG THE MAIN DISRUPTION DRIVERS
# ============================================================

recommendation_drivers["critical_part_exposure"] = (
    recommendation_drivers["critical_parts_affected"]
    >= critical_parts_threshold
)

recommendation_drivers["high_procurement_exposure"] = (
    recommendation_drivers["procurement_value_exposed"]
    >= procurement_threshold
)

recommendation_drivers["high_consumption_exposure"] = (
    recommendation_drivers["average_daily_consumption"]
    >= consumption_threshold
)

recommendation_drivers["high_blocked_quantity"] = (
    recommendation_drivers["total_blocked_quantity"]
    >= blocked_threshold
)

recommendation_drivers["high_backorder"] = (
    recommendation_drivers["total_backorder"]
    >= backorder_threshold
)

recommendation_drivers["high_disruption_exposure"] = (
    recommendation_drivers["disruption_exposure_rank"] <= 5
)

recommendation_drivers["high_supplier_risk"] = (
    recommendation_drivers["risk_category"] == "Critical"
)
```

```python
recommendation_flags = recommendation_drivers[
    [
        "supplier_id",
        "disruption_exposure_rank",
        "risk_category",
        "supplier_priority_score",
        "critical_parts_affected",
        "procurement_value_exposed",
        "average_daily_consumption",
        "total_backorder",
        "total_blocked_quantity",
        "single_source_parts",
        "critical_part_exposure",
        "high_procurement_exposure",
        "high_consumption_exposure",
        "high_blocked_quantity",
        "high_backorder",
        "high_disruption_exposure",
        "high_supplier_risk"
    ]
].sort_values(
    "disruption_exposure_rank"
)

display(recommendation_flags)
```

```python
# ============================================================
# SUPPLIER-SPECIFIC RECOMMENDATIONS
# ============================================================

def generate_recommendation(row):

    # Highest priority: critical parts + major disruption exposure
    if (
        row["high_disruption_exposure"]
        and row["critical_part_exposure"]
    ):
        return (
            "Immediate alternate-source qualification for critical parts",
            "Immediate",
            "High disruption exposure combined with multiple critical parts and single-source dependence."
        )

    # Critical parts even when overall disruption rank is lower
    elif row["critical_part_exposure"]:
        return (
            "Qualify alternate suppliers for critical parts",
            "High",
            "Multiple critical parts are exposed and the affected parts are single-source."
        )

    # Large operational demand exposure
    elif (
        row["high_disruption_exposure"]
        and row["high_consumption_exposure"]
    ):
        return (
            "Develop inventory and supply continuity contingency plan",
            "High",
            "High disruption exposure coincides with high daily consumption."
        )

    # Significant blocked quantity / backorders
    elif (
        row["high_blocked_quantity"]
        or row["high_backorder"]
    ):
        return (
            "Initiate supplier corrective action and operational recovery review",
            "High",
            "Elevated blocked quantity or backorder exposure indicates operational disruption."
        )

    # Large procurement exposure
    elif row["high_procurement_exposure"]:
        return (
            "Review procurement concentration and supply continuity options",
            "Medium",
            "High procurement value is exposed if the supplier becomes unavailable."
        )

    # High supplier risk but relatively low disruption consequence
    elif row["high_supplier_risk"]:
        return (
            "Enhanced supplier monitoring and periodic risk review",
            "Medium",
            "Supplier is classified as Critical risk, although disruption exposure is comparatively lower."
        )

    else:
        return (
            "Routine supplier monitoring",
            "Low",
            "No dominant disruption driver identified relative to the other suppliers."
        )


# Apply recommendation logic
recommendation_results = (
    recommendation_drivers.apply(
        generate_recommendation,
        axis=1,
        result_type="expand"
    )
)

recommendation_results.columns = [
    "recommended_action",
    "recommendation_urgency",
    "recommendation_rationale"
]

# Attach recommendations
supplier_recommendations = pd.concat(
    [
        recommendation_drivers.reset_index(drop=True),
        recommendation_results.reset_index(drop=True)
    ],
    axis=1
)

# Final business-facing table
supplier_recommendations = supplier_recommendations[
    [
        "supplier_id",
        "disruption_exposure_rank",
        "risk_category",
        "supplier_priority_score",
        "affected_parts",
        "critical_parts_affected",
        "single_source_parts",
        "procurement_value_exposed",
        "critical_procurement_value_exposed",
        "average_daily_consumption",
        "total_backorder",
        "total_blocked_quantity",
        "recommended_action",
        "recommendation_urgency",
        "recommendation_rationale"
    ]
].sort_values(
    "disruption_exposure_rank"
)

display(supplier_recommendations)
```

```python
# ============================================================
# SAVE FINAL SUPPLIER RECOMMENDATIONS
# ============================================================

recommendation_output = supplier_recommendations.copy()

output_path = "data/processed/supplier_recommendations.csv"

recommendation_output.to_csv(
    output_path,
    index=False
)

print("Saved successfully:")
print(output_path)

print("\nFinal recommendation table shape:")
print(recommendation_output.shape)

display(recommendation_output)
```

```python
# Validate recommendation distribution

print("Recommendation urgency:")
print(
    recommendation_output["recommendation_urgency"]
    .value_counts()
)

print("\nRecommended actions:")
print(
    recommendation_output["recommended_action"]
    .value_counts()
)
```

```python
# ============================================================
# EXECUTIVE KPI SUMMARY
# ============================================================

kpi_summary = pd.DataFrame({
    "metric": [
        "Total Suppliers Analysed",
        "Critical-Risk Suppliers",
        "High-Risk Suppliers",
        "Total Affected Parts",
        "Total Critical Parts Affected",
        "Total Order Quantity Exposed",
        "Total Procurement Value Exposed",
        "Total Critical Procurement Value Exposed",
        "Total Backorder Exposure",
        "Total Blocked Quantity",
        "Single-Source Parts"
    ],
    
    "value": [
        disruption["supplier_id"].nunique(),
        (disruption["risk_category"] == "Critical").sum(),
        (disruption["risk_category"] == "High").sum(),
        disruption["affected_parts"].sum(),
        disruption["critical_parts_affected"].sum(),
        disruption["total_order_quantity_exposed"].sum(),
        disruption["procurement_value_exposed"].sum(),
        disruption["critical_procurement_value_exposed"].sum(),
        disruption["total_backorder"].sum(),
        disruption["total_blocked_quantity"].sum(),
        disruption["single_source_parts"].sum()
    ]
})

display(kpi_summary)
```

```python
# ============================================================
# TOP DISRUPTION SCENARIOS
# ============================================================

top_disruption_suppliers = (
    disruption
    .sort_values("disruption_exposure_rank")
    .head(10)
    [
        [
            "disruption_exposure_rank",
            "supplier_id",
            "risk_category",
            "supplier_priority_score",
            "affected_parts",
            "critical_parts_affected",
            "procurement_value_exposed",
            "critical_procurement_value_exposed",
            "average_daily_consumption",
            "total_backorder",
            "total_blocked_quantity"
        ]
    ]
)

display(top_disruption_suppliers)
```

```python
# ============================================================
# MANAGEMENT ACTION SUMMARY
# ============================================================

management_actions = (
    supplier_recommendations
    [
        [
            "supplier_id",
            "disruption_exposure_rank",
            "risk_category",
            "recommended_action",
            "recommendation_urgency",
            "recommendation_rationale"
        ]
    ]
    .sort_values(
        ["recommendation_urgency", "disruption_exposure_rank"],
        ascending=[True, True]
    )
)

display(management_actions)
```

```python
# ============================================================
# SAVE DASHBOARD-READY OUTPUTS
# ============================================================

kpi_summary.to_csv(
    "data/processed/executive_kpi_summary.csv",
    index=False
)

top_disruption_suppliers.to_csv(
    "data/processed/top_disruption_suppliers.csv",
    index=False
)

management_actions.to_csv(
    "data/processed/management_actions.csv",
    index=False
)

print("Dashboard-ready files saved successfully:")
print("1. data/processed/executive_kpi_summary.csv")
print("2. data/processed/top_disruption_suppliers.csv")
print("3. data/processed/management_actions.csv")
```
